### Escape sequences

A backslash `\` inside a string starts an **escape sequence**: a special character that is awkward to type.

| Escape | Meaning |
|---|---|
| `\n` | new line |
| `\t` | tab |
| `\\` | a literal backslash |
| `\'` | a literal single quote |
| `\"` | a literal double quote |

````python
print("Line one\nLine two")
print("Name:\tAda")
print("She said \"hello\"")
print("A backslash: \\")
````

### Raw strings

Sometimes you don't want escapes: Windows file paths are the classic case, because they are full of backslashes. Put an `r` before the opening quote to make a **raw string**, where backslashes are just backslashes:

````python
print("C:\new\table")      # \n and \t were treated as escapes!
print(r"C:\new\table")     # raw string: exactly what we typed
````

### Showing a string versus printing it

When a notebook displays a string as the last line of a cell, it shows the quotes and any escapes. `print()` shows the text as a human would read it:

````python
"first line\nsecond line"
````

````python
print("first line\nsecond line")
````

Both are the same string; there are just two ways of *looking* at it. The first form (called the `repr`) is useful for debugging, because it shows exactly what's in there, including invisible characters like newlines and tabs.

## 4 · Indexing: picking out one character

Each character has a **position**, called its **index**. Counting starts at **0**, not 1:

```
 string:    P   y   t   h   o   n
 index:     0   1   2   3   4   5
 negative: -6  -5  -4  -3  -2  -1
```

Write the index in square brackets after the string:

````python
word = "Python"
print(word[0])
print(word[1])
print(word[2])
````

**Negative indices count from the end:** `-1` is the last character, `-2` the second to last, and so on. Handy, because you don't need to know the length.

````python
print(word[-6])
print(word[-5])
````

````python
word=("yash")
print(word[-4])
````

The valid indices for a string of length `n` are `0` to `n - 1` (or `-1` down to `-n`). Anything else is an error. ⚠️ **Expected to fail.**

````python
print(word[4])
````

A single character is just a one-character string; Python has no separate "character" type.

````python
print(type(word[0]))
````

## 5 · Slicing: picking out a piece

A **slice** takes a piece of a string: `text[start:stop]`. It includes `start` but stops **just before** `stop`.

A good way to remember this: imagine the indices point to the *gaps between* characters:

```
   0   1   2   3   4   5   6
   | P | y | t | h | o | n |
```

So `word[1:4]` means "from gap 1 to gap 4", which is `yth`.

````python
word = "Python"
print(word[1:4])
print(word[0:2])
print(word[2:6])
````

`stop - start` is the length of the slice. That's a nice way to double check yourself: `word[1:4]` has length `4 - 1 = 3`.

### Leaving things out

If you omit `start`, the slice starts at the beginning. If you omit `stop`, it goes to the end.

````python
print(word[:3])     # first three characters
print(word[3:])     # everything from index 3 onwards
print(word[:])      # a copy of the whole string
````

A handy fact: `word[:n] + word[n:]` is always the original string.

### Negative indices in slices

````python
print(word[-3:])     # last three characters
print(word[:-1])     # everything except the last character
print(word[1:-1])    # remove first and last
````

### Steps: `text[start:stop:step]`

A third number says how many to skip each time:

````python
digits = "0123456789"
print(digits[::2])      # every second character
print(digits[1::2])     # every second, starting at index 1
print(digits[::3])
````

A negative step walks **backwards**. The most famous slice in Python reverses a string:

````python
#python
print(digits[::-1])
print(word[::-1])
````

### Slices are forgiving

Unlike a single index, a slice never raises an error when it goes past the ends. It just gives you whatever is there:

````python
print(word[2:100])
print(word[10:20] == "")
````

````python
word="python"
print(word[0:6])
print(word[0:6] =="python")
````

## 6 · Strings are immutable

You can read any character but you **can't change one in place**. ⚠️ **Expected to fail.**

````python
word = "Python"
word[0] =str("P")
````

This is called **immutability**: a string, once created, never changes. If you want a different string, you **build a new one** (and optionally store it under the same name):

````python
word = "Python"
new_word = "J" + word[1:]
print(new_word)
print(word)         # the original is untouched
````

Immutability is a feature: it means a string can never change behind your back. Everything that looks like it "modifies" a string (and in notebook 05 we'll meet lots of those) really produces a new one.

## 7 · Strings are sequences, and the `in` operator

Because a string is a sequence of characters, it supports some other handy operations. The `in` operator checks for a **substring**:

````python
print("thon"not  in "Python")
print("z" in "Python")
print("Py" in "Python")        # case matters!
print("a" not in "Python")
````

The answers `True` and `False` are *booleans*; notebook 07 is all about them. You'll use `in` with `if` in notebook 08.

## 8 · Common mistakes

### 1. Adding a number to a string

`+` can add two numbers or join two strings, but not a mix. ⚠️ **Expected to fail.**

````python
age = 30
print("I am " , age , " years old")
````

Fixes: use commas in `print()` (`print("I am", age, "years old")`), or convert the number to text with `str(age)`, which we'll study properly in notebook 06.

````python
age = 30
print("I am " + str(age) + " years old")
print("I am", age, "years old")
````

### 2. Off-by-one in slices

`word[1:4]` ends **before** index 4. If you expect four characters from index 1, you want `word[1:5]`. Rule: the length of `text[a:b]` is `b - a`.

### 3. Counting from 1

The first character is `text[0]`, not `text[1]`. The last is `text[len(text) - 1]` (or just `text[-1]`).

### 4. Forgetting `len` needs brackets

`len` on its own is the function; `len(text)` is the answer.

### 5. Multiplying a string by a decimal

`"ab" * 2` is fine; `"ab" * 2.5` is not. ⚠️ **Expected to fail.**

````python
print("ab" * 2)
````

### 6. Mismatched or "curly" quotes, and unescaped backslashes

Remember notebook 01: the opening and closing quotes must match, and they must be straight. A Windows path like `"C:\temp\new"` hides escape sequences (`\t` and `\n`!). Use a raw string.

### 7. Expecting a string to change

Writing `word[0] = "J"` fails; writing `word + "!"` on its own line does nothing useful, since it creates a new string and throws it away unless you store it:

````python
word = "Python"
word + "!"          # computes a new string, and then it is discarded
print(word)
word = word + "!"   # now we keep it
print(word)
````

## 10 · Recap and what's next

- Strings are made with `'single'`, `"double"` or `"""triple"""` quotes. Escapes such as `\n`, `\t`, `\\` insert special characters; raw strings `r"..."` ignore them.
- `+` joins, `*` repeats, `len()` measures. You can't `+` a string and a number.
- **Index** from 0 (or from the end with negatives): `s[0]`, `s[-1]`.
- **Slice** with `s[start:stop:step]`: `start` included, `stop` excluded; omit any part; a step of `-1` reverses.
- Strings are **immutable**: build new ones rather than changing old ones.
- `in` tests whether one string is inside another.
